```python
from deep_translator import (GoogleTranslator,
                             ChatGptTranslator,
                             MicrosoftTranslator,
                             PonsTranslator,
                             LingueeTranslator,
                             MyMemoryTranslator,
                             YandexTranslator,
                             PapagoTranslator,
                             DeeplTranslator,
                             QcriTranslator)
import pandas as pd

# for translator in [ GoogleTranslator, ChatGptTranslator, 
#     MicrosoftTranslator, PonsTranslator, LingueeTranslator,
#     MyMemoryTranslator, YandexTranslator, PapagoTranslator,
#     DeeplTranslator, QcriTranslator]:
#     print()
#     print(translator.__name__)
#     try:
#         print(translator().get_supported_languages())
#     except Exception as e:
#         print(e)


sentences = [
    'The cat slept peacefully under the warm sunlight.',
    'He finds a strange note hidden in the drawer.',
    'We watched the stars until morning slowly arrived.',
    'A door is opened at the back of the house.',
    'Her laughter echoed through the empty, quiet hallway.',
    'The old bridge creaked beneath the heavy wagon.',
    'They discovered fresh footprints in the wet sand.',
    'I will speak to him first thing tomorrow morning.',
    'I forgot my umbrella during the sudden storm.',
    'Will you walk with me to the train station?',
    'The train vanishes into the dark mountain tunnel.',
    'A small bird perched gently on the windowsill.',
    'She whispered a secret nobody else could hear.',
    'The lady waits with a newspaper in her hand.',
    'You won\'t believe what I heard yesterday.',
    'He\'s going to sing in the concert this evening.'
]

data = {
    'Original sentence': sentences,
    'Google Translate': [GoogleTranslator("en", "cy").translate(sentence) for sentence in sentences],
    'My Memory': [MyMemoryTranslator(source="en-GB",target="cy-GB").translate(sentence) for sentence in sentences]
}

df = pd.DataFrame(data)

df
```

```python
df.to_csv('translation_comparison.csv')
```
